## Strategy and reasoning

This is the final selected agent, not an experimental rewrite. Ten historical and current lineages were screened on 32 recent live-ladder opponent action streams using the original seeds and seats. The four finalists were then confirmed on 24 separate historical high-band opponent streams. Selection used win/loss/tie points first, catastrophic-loss count second, and coin margin only as a tertiary diagnostic.

V94 restores the V81 Spatial Mirror Gate. It retains the complete stable production policy and changes market timing only when both public farms have a coordinate-level matching fingerprint and a small cash gap. In those likely same-lineage races it uses the seven-turn sale lead; otherwise it keeps the conservative four-turn horizon. It uses no hidden information, opponent identity, or replay fingerprint.
